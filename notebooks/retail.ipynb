```python
import pandas as pd
print(pd.__version__)
```

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine

load_dotenv()

DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")

DATABASE_URL = (
    f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}"
    f"@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

engine = create_engine(DATABASE_URL)

df = pd.read_sql(
    "SELECT * FROM customer_analytics;",
    engine
)

df.head()
```

```python
print(df.shape)
print(df.columns)
print(df.isna().sum())
```

```python

print(df['favorite_category'].unique())

df['favorite_category']= df['favorite_category'].fillna(df['favorite_category'].mode()[0])
print(df.isna().sum())
```

```python
print(df['first_order_date'].describe())
print(df['last_order_date'].describe())

df[df["first_order_date"].isna()][
    ["customer_id", "total_orders", "total_spending"]
].head()

df["has_purchase"] = (df["total_orders"] > 0).astype(int)
```

```python

print(df.dtypes)
```

```python
print(df['gender'].unique())
print(df['city'].unique())
print(df['state'].unique())
print(df['occupation'].unique())
print(df['marital_status'].unique())
print(df['preferred_channel'].unique())
print(df['loyalty_tier'].unique())
print(df['favorite_category'].unique())
print(df['customer_value_segment'].unique())
```

```python
df = pd.get_dummies(
    df,columns=["gender","marital_status","state","occupation",
                "favorite_category","preferred_channel"
    ],dtype=int)
```

```python
#frequcnecy coz the number of cities are high (too much dummy cols)
city_frequency = df["city"].value_counts(normalize=True)
df["city_frequency"] = df["city"].map(city_frequency)
df.drop(columns=["city"], inplace=True)

#ordininal coz it has order
loyalty_map = {
    "Bronze": 1,
    "Silver": 2,
    "Gold": 3,
    "Platinum": 4
}

df["loyalty_tier"] = df["loyalty_tier"].map(loyalty_map)
df.drop(columns=["customer_value_segment"], inplace=True)
```

```python
df.isna().sum()
```

```python
df["first_order_date"] = pd.to_datetime(df["first_order_date"])
df["last_order_date"] = pd.to_datetime(df["last_order_date"])
df["customer_tenure_days"] = (
    df["last_order_date"] - df["first_order_date"]
).dt.days



reference_date = df["last_order_date"].max()
df["recency"] = (
    reference_date - df["last_order_date"]
).dt.days
print(reference_date)
df["frequency"] = df["total_orders"]
df["monetary"] = df["total_spending"]
```

```python
rfm = df[df["has_purchase"] == 1][
    ["customer_id", "recency", "frequency", "monetary"]
].copy()

print(rfm.shape)
print(rfm.isna().sum())
```

```python
from sklearn.preprocessing import PowerTransformer

pt = PowerTransformer(method="yeo-johnson")

rfm_transformed = pt.fit_transform(
    rfm[["recency", "frequency", "monetary"]]
)

rfm_transformed = pd.DataFrame(
    rfm_transformed,
    columns=["recency", "frequency", "monetary"],
    index=rfm.index
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler= StandardScaler()

rfm_scaled= scaler.fit_transform(rfm_transformed)

rfm_scaled =pd.DataFrame(rfm_scaled, columns=['recency','frequency','monetary'], index=rfm.index)

print(rfm_scaled.describe())
    
```

```python
print(rfm_scaled.shape)
print(rfm_scaled.isna().sum())
```

```python
from sklearn.decomposition import PCA

pca = PCA()

rfm_pca = pca.fit_transform(rfm_scaled)

print("Explained variance:")
print(pca.explained_variance_ratio_)

print("Cumulative variance:")
print(pca.explained_variance_ratio_.cumsum())
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

rfm_pca = pca.fit_transform(rfm_scaled)

rfm_pca = pd.DataFrame(
    rfm_pca,
    columns=["PC1", "PC2"],
    index=rfm.index
)

print(rfm_pca.head())
print("Explained variance:", pca.explained_variance_ratio_)
print("Total variance:", pca.explained_variance_ratio_.sum())
```
